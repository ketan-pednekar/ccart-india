```python
import xarray as xr
import rioxarray

# Path to your CHIRPS India Zarr
zarr_path = r"C:\CMIP_data\cmip6\Climada\Projects\ccart-india\ccart\outputs\chirps_india\chirps_india_clipped.zarr"

# Load dataset
ds = xr.open_zarr(zarr_path)

# Select any time slice (grid only)
da = ds["pr"].isel(time=0)

# Tell rioxarray which dims are spatial
da = da.rio.set_spatial_dims(x_dim="lon", y_dim="lat", inplace=False)

# Ensure CRS is attached
da = da.rio.write_crs("EPSG:4326")

# Output path
out_tif = r"C:\CMIP_data\cmip6\Climada\Projects\ccart-india\ccart\outputs\chirps_india\chirps_india_template.tif"

# Save raster
da.rio.to_raster(out_tif)

print("Saved:", out_tif)


```
