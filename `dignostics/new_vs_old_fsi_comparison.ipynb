```python
import rasterio
import numpy as np

old_path = r"C:\CMIP_data\cmip6\Climada\Projects\ccart-india\ccart\flood\outputs\ccart_floods_fsi_static_chirps_rescaled.tif"
new_path = r"C:\CMIP_data\cmip6\Climada\Projects\ccart-india\ccart\flood\outputs\fsi\ccart_floods_fsi_static_chirps_rescaled.tif"

def load_raster(path):
    with rasterio.open(path) as src:
        arr = src.read(1)
        return arr, src.transform, src.crs, src.shape

old_arr, old_transform, old_crs, old_shape = load_raster(old_path)
new_arr, new_transform, new_crs, new_shape = load_raster(new_path)

print("\n=== STRUCTURAL CHECKS ===")
print("Same shape:", old_shape == new_shape)
print("Old shape:", old_shape)
print("New shape:", new_shape)

print("\nSame CRS:", old_crs == new_crs)
print("Old CRS:", old_crs)
print("New CRS:", new_crs)

print("\nSame transform:", old_transform == new_transform)
print("Old transform:", old_transform)
print("New transform:", new_transform)

print("\n=== VALUE DIFFERENCE CHECK ===")
# Mask NaNs for safe comparison
mask = ~np.isnan(old_arr) & ~np.isnan(new_arr)

if mask.sum() == 0:
    print("No overlapping valid pixels to compare.")
else:
    diff = new_arr[mask] - old_arr[mask]
    print("All equal:", np.allclose(new_arr[mask], old_arr[mask], equal_nan=True))
    print("Max abs diff:", np.nanmax(np.abs(diff)))
    print("Mean diff:", np.nanmean(diff))
    print("Min diff:", np.nanmin(diff))
    print("Max diff:", np.nanmax(diff))

print("\n=== NAN CHECK ===")
print("Old NaNs:", np.isnan(old_arr).sum())
print("New NaNs:", np.isnan(new_arr).sum())
```
