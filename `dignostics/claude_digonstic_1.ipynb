```python
import geopandas as gpd
import numpy as np

# Check the FSI output
from ccart.flood.fsi.compute_fsi import compute_fsi

gdf = compute_fsi()

print(f"Total gauges          : {len(gdf)}")
print(f"FSI_v1_1 range        : {gdf['FSI_v1_1'].min():.4f} – {gdf['FSI_v1_1'].max():.4f}")
print(f"FSI_v1_2 range        : {gdf['FSI_v1_2'].min():.4f} – {gdf['FSI_v1_2'].max():.4f}")
print(f"FSI_masked valid      : {gdf['FSI_masked'].notna().sum()}")
print(f"FSI_masked NaN (proxy): {gdf['FSI_masked'].isna().sum()}")
print(f"Proxy_flag == 1       : {(gdf['Proxy_flag'] == 1).sum()}")
print(f"\nBasins represented    : {gdf['Basin'].nunique()}")
print(f"Basin coverage:")
print(gdf.groupby('Basin')['FSI_masked'].agg(['count', 'mean']).round(4))
```
